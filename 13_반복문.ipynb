## for문 실습 03

[문제] 아래의 실행 결과처럼 나오도록 for문과 range()를 활용해서 구현해봅시다. \
- 숫자와 숫자 사이에 \t를 사용 \

[실행 결과] \
자연수 하나 입력해주세요: 6 \
6 6 6 6 6 6 \
5 5 5 5 5 \
4 4 4 4 \
3 3 3 \
2 2 \
1 

```python
n = int(input("자연수 하나 입력해주세요: "))

for i in range(n, 0, -1):  # n부터 1까지 반복
    print("\t".join([str(i)] * i))  # 숫자 i를 i번 반복하여 탭으로 구분해서 출력
```

```python
n = int(input("자연수 하나 입력해주세요: "))

for i in range(n, 0, -1): 
    for j in range(i):
        print(i, end="\t")  
    print()  
```

## for문 실습 05

[문제] 1부터 사용자로부터 입렵 받은 자연수까지의 홀수의 합을 구합니다.
- 1부터 입력받은 수까지 홀수 전체를 각각 더하는 산식을 표현하고 
- 그 계산식의 결과를 아래 실행 결과와 같이 출력되도록 구해보세요

[실행 결과] \
1부터 입력한 자연수까지의 모든 홀수의 합을 구합니다: 100 \
1 + 3 + 5 + 7 + 9 + … + 99 = 2500

```python
a = int(input('자연수를 입력해 주세요.: '))
print('1부터 입력한 자연수 까지의 모든 홀수의 합을 구합니다: ', a)
result = 0

for i in range(1,a+1):
    if i%2 ==1:
        if i == a or i == (a-1):
            print(i,end=' = ')
        else:
            print(i, end= ' + ')
        result += i

print(result)
```

## for문 실습 06

[문제] 계승(Factorial)을 계산해보려 합니다. 
- 사용자로부터 정수 하나를 입력 받습니다.
- 입력 받은 숫자까지의 계승을 for문을 활용하여 구한 뒤 출력해주세요.
- 계승이란 1부터 n까지의 자연수를 모두 곱하는 것을 의미합니다. 
- n! = 1 x 2 x 3 x … x (n-1) x n


[실행 결과] \
계승을 구할 숫자를 입력하세요: 4 \
24 

계승을 구할 숫자를 입력하세요: 7 \
5040 



```python
a = int( input())
print('계승을 구할 숫자를 입력하세요.: ', a)
result = 1

for i in range(1,a+1):
    result *= i

print(result)
```

# 3. while문 

## 3.1 while문 작성해보기

```python
i = 0
while i < 5:
    print(f'Hello, World! {i}')
    i += 1
```

- 입렵한 횟수대로 반복하는 예제
- 다만, 숫자를 감소시켜 제어

Hello, World 3 \
Hello, World 2 \
Hello, World 1 

```python
count = int(input())

while count > 0 :
    print('Hello, World', count)
    count -= 1
```
