## 1.2 함수정의 실습 01

```python
# 방법 1
def odd_even(number):
    """number가 짝수면 '짝수', 홀수면 '홀수' 반환"""

    if type(number) is int and number > 0:
        return '짝수' * (number % 2 == 0) or '홀수'
    else:
        return
```

```python
"""
'짝수' * (number % 2 ==0):
    - '짝수' * True -> '짝수' * 1 -> '짝수'
    - '짝수' * False -> '짝수' * 0 -> ''(빈 문자열)

or '홀수':
    - '짝수' * (number % 2 == 0)가 빈 문자열이면 '홀수'반환
    - or 연산자는 앞쪽 피연산자가 '참'으로 평가되면 그 값을 반환하고, '거짓'으로 평가되면 뒤쪽 피연산자를 반환 
    - 즉, 빈 문자열은 '거짓'으로 평가되고, '홀수'는 '참'으로 평가 됨 
즉, 조건에 따라 '짝수' 또는 '홀수'가 반환 됨
"""
```

```python
print(odd_even.__doc__)
print(odd_even(10))
print(odd_even(9))
print(odd_even('가'))
```

```python
# 방법 2 

def odd_even(number):
    """number가 짝수면 '짝수', 홀수면 '홀수' 반환"""

    if type(number) is int and number > 0:
        if number % 2 == 0:
            return '짝수'
        else:
            return '홀수'
    else:
        return
```

```python
print(odd_even.__doc__)
print(odd_even(10))
print(odd_even(9))
print(odd_even('가'))
```

## 1.3 함수정의 실습 02

```python
def check_leap_year(y):
    if y % 4 == 0 and (y % 100 != 0 or y % 400 == 0): # 100으로 나눠지지 않거나 400으로 나눠져야함
        return '윤년'
    else:
        return '평년'
```

```python
print(check_leap_year(2004))
print(check_leap_year(2000))
print(check_leap_year(1900))
```

## 2.5 가변-위치 인자 (var-positional parameters)

1. 가변-위치 인자에 위치 인수 전달 

```python
# 함수 정의
def var_positional(*test):	
	print(type(test))
	return test

# 함수 호출
print(var_positional(1, 2, 3)) # 위치 인수들이 튜플로 패킹되어 함수 인자에 전달됨을 알 수 있다.
```

```python
# 함수 정의
def var_positional(*args):	
	print(type(args))
	return [i for i in args] # 전달된 튜플이 for문 통해 언패킹 됨 -> list로 변환됨

# 함수 호출
print(var_positional(1, 2, 3))
```

```python
important_is_an_unbroken_heart = '중요한 것은 꺾이지 않는 마음' # 전역변수

def trials_and_tribulations():
    important_is_an_unbroken_heart = '흔들흔들' # 지역 범위에서 값 변경 시도


trials_and_tribulations()
print(important_is_an_unbroken_heart) # 값 변경이 되지 않음을 알 수 있음 
```

# 4. 중첩 함수

## 4.1 중첩 함수 만들어보기

```python
def level_1():
    message = "This is level 1"
    def level_2():
        print(message)
    level_2() # 중첩 함수를 나와서 중첩 함수를 한 번 호출 해줘야 함 


level_1()
```

## 4.2 안쪽 함수에서 바깥쪽 함수의 변수 변경해보기

```python
def level_1():
    message = "This is level 1"
    def level_2():
        nonlocal message
        message = "Level 2 is better than level 1"
    level_2() # 중첩 함수를 나와서 중첩 함수를 한 번 호출 해줘야 함 
    print(message)


level_1()
```

## 4.3 nonlocal 주의점
1. 중첩된 함수마다 같은 이름의 변수가 있다면 nonlocal 키워드는 제일 가까운 바깥 변수 매칭


```python
def level_1():
    message = "This is level 1"

    def level_2():
        message = "Level 2 is better than level 1"

        def level_3():
            nonlocal message # 제일 가까운 message가 매칭됨
            print(message) # level_2의 message가 print 됨
        level_3()

    level_2()
    


level_1()
```

2. 파이썬 자체에서 정의한 이름 공간을 내장 영역(built-in scope)이라고 함
- 함수에서 변수를 호출하면 지역 영역 -> 전역 영역 -> 내장 영역 순으로 해당하는 변수를 확인
- 즉, 아래의 예시처럼 전역 영역에서 len을 재정의하면 내장 len함수가 가려짐 
- 하지만, del로 전역 len을 삭제하면 내장 여역에 있는 len이 다시 사용됨


```python
len = 100  # 전역 영역에 len 이름 정의
print(len)  # 출력: 100 (전역 len 사용)

del len  # 전역 len 제거
print(len("Hello"))  # 출력: 5 (내장 len 사용)
```
