# 1. 인스턴스 (Instance)
인스턴스와 객체
- 클래스(Class)를 기반으로 생성된 구체적인 객체를 의미
- 즉, 인스턴스 = 클래스이름()에서 만들어진 객체


```python
class Dog:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
puppy1 = Dog('Buddy', 3)
puppy2 = Dog('Max', 5)

print('name = ', puppy1.name)
print('age = ', puppy1.age)

print('name = ', puppy2.name)
print('age = ', puppy2.age)
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name}가 짖습니다!")
    
# Dog 객체 생성
dog1 = Dog("바둑이", 5)
dog2 = Dog("흰둥이", 3)

# 객체 속성 출력
print(dog1.name)  
print(dog2.age) 
```

# 2. 메서드 (Method)

- 클래스 바디 안에 정의되는 함수 
- 클래스의 인스턴스 속성으로 호출되면, 그 메서드는 첫 번째 인자로 인스턴스 객체를 받음
- 첫 번째 인자를 설정 안 하면 에러 발생
- 이 첫 번째 인자를 'self'라고 씀

```python
class Person:
    def __init__(self):
        self.hello = '안녕하세요'

    def greeting(self):
        print('hi')

james = Person()
james.greeting()
```

# 3. 클래스 속성 (Attribute)
- 모든 객체(인스턴스)가 공유하는 속성
- 클래스 메소드에서 객체 없이 클래스명으로 접근 가능


```python
class Dog:

    dog_count = 0 # 클래스 속성

    def __init__(self, name):
        self.name = name 
        Dog.dog_count += 1
```

```python
puppy1 = Dog('Buddy')
print(puppy1.name)
print(puppy1.dog_count)
```

```python
puppy2 = Dog('Max')
print(puppy2.name)
print(puppy2.dog_count)
```

# 8. 비공개 속성 사용하기 

```python
class Person:

    def __init__(self, name, age, address,wallet):
        self.name = name
        self.age = age
        self.address = address
        self.__wallet = wallet
    def pay(self,amount):
        if self.__wallet < amount:
            print('돈이 모자랍니다.')
        self.__wallet -= amount
        print('이제 {0}원 남았네요.'.format(self.__wallet))
```

```python
maria=Person('마리아',20,'서울시 서초구 반포동',10000)
maria.pay(3000)
maria.pay(10000)
```

```python
class Person:

    def __init__(self, name, age, address,wallet):
        self.name = name
        self.age = age
        self.address = address
        self.__wallet = wallet
    def pay(self,amount):
        if self.__wallet < amount:
            print('돈이 모자랍니다.')
        else:
            self.__wallet -= amount
            print(f'이제 {self.__wallet}원 남았네요.')
```

1. 예제1

```python
def logger(func):
    def wrapper(*args,**kwargs):
        print('함수 실행중인 함수: {func.__name__}')
        return func(*args,**kwargs)
    return wrapper
```

```python
@logger
def say_hello(name):
    print(f'안녕하세요, {name}님!')


say_hello('철수')
```

```python
def say_hello(name):
    print(f'{name}')
say_hello.__name__
```

```python
class MyClass:
    class_variable = 0

    @classmethod
    def increment_class_variable(cls):
        cls.class_variable += 1
        print(f'클래스 변수값: {cls.class_variable}')

MyClass.increment_class_variable()

MyClass.increment_class_variable()
```
