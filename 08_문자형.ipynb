# 6. 서식지정자

## 6.1 %s 포매팅

1. 기본 사용 법

```python
# %s 포메팅 사용 안한 경우
name = 'tom'

print('I am '+name+'!')
```

```python
# %s 포메팅 사용한 경우
name = 'sam'

print('I am %s!'%name)
```

```python
name = input()

print('I am %s'%name + '!')
```

2. 응용1: 공백 추가

```python
# 왼쪽 공백 추가
name = 'sam'

print('I am %10s'%name+'!')
```

```python
# 오른쪽 공백 추가
name = 'sam'

print('I am %-10s!'%name)
```

3. 여러개의 문자 추가

```python
string1='an apple'
string2='a banana'

print('I like %s and %s!' %(string1, string2))
```

## 6.2 %d 포매팅 

```python
number = 5
print('Number %d is my favorite number.'%number)
```

```python
test = 'Number %s is my favorite number.'%number
print(type(test))

test2 = 'Number %d is my favorite number.'%number
print(type(test2))
```

## 6.3 %f 포매팅 

1. 기본 사용 방식

```python
number = 5.4
print('Number %d is my favorite number.'%number)

number = 5.4
print('Number %f is my favorite number.'%number)

```

2. 응용1: 지정한 소수점 이하 자리까지 출력

```python
number = 5.4
print('Number %f is my favorite number.'%number)

number = 5.4
print('Number %.2f is my favorite number.'%number)
```

3. 응용2: 너비만큼 0으로 채우기

```python
num = 3.23
print('with %012f: %012.2f' %(num, num))
```

# 7. format 함수

1. 기본 사용 방식

```python
# "문자열 {0}, {1} 문자열".format(값1,값2)
print('I like {0} and {1}.'.format("an apple","a banana"))
print('I like {1} and {0}.'.format("an apple","a banana"))
print('I like {} and {}.'.format("an apple","a banana"))
```

2. 변수 활용

```python
fruit1, fruit2 = 'an apple', 'a banana'

print('I like {0} and {1}.'.format(fruit1, fruit2))
```

```python
test1 = 'Numeber {0}, {1}, {2}'.format(4,5,6)
test2 = 'Numeber {0}, {0}, {1}'.format(4,5,6)
test3 = 'Numeber {2}, {1}, {0}'.format(4,5,6)

print('test1: ',test1)
print('test2: ',test2)
print('test3: ',test3)
```

## 05-2 

[문제] 이름과 점수 3개를 입력받아 아래와 같이 출력하세요


[실행 결과] \
이름을 입력해주세요: 홍길동 \
첫 번째 점수를 입력해주세요: 80 \
두 번째 점수를 입력해주세요: 70 \
세 번째 점수를 입력해주세요: 60 \
저의 이름은 홍길동이고 총점은 210 입니다.


```python
name = input()
score1 = 80
score2 = 70
score3 = 60

print(f'이름을 입력해주세요: {name}')
print(f'첫 번째 점수를 입력해주세요.: {score1}')
print(f'두 번째 점수를 입력해주세요.: {score2}')
print(f'세 번째 점수를 입력해주세요.: {score3}')
print(f'저의 이름은 {name}이고, 총점은 {score1+score2+score3}입니다.')
```

```python
# 이름과 점수 3개 입력 받기
name = input("이름을 입력해주세요: ")
score1 = int(input("첫 번째 점수를 입력해주세요: "))
score2 = int(input("두 번째 점수를 입력해주세요: "))
score3 = int(input("세 번째 점수를 입력해주세요: "))

# 총점 계산
sum_ = score1 + score2 + score3


print(f"저의 이름은 {name}이고 총점은 {sum_} 입니다.")
```
