# 1. 자료형 확인
- type 클래스 (`type()`)를 통해 데이터의 유형(타입)을 확인할 수 있다.

## 1.1 int 타입 확인

```python
type('hello')
```

```python
"hello\n world"
print("hello\n world")
```

```python
print([1, 2, 3])
1.1
print(type(1))
print(type(2))
type(3.1)
```

## 1.2 sequence 타입 확인 

### 1.2.1 str 타입

```python
print(type('Hello, World!'))
print(type('1'))
```

### 1.2.2 list 타입

```python
print(type([1,2,3]))
```

### 1.2.3 tuple 타입

```python
print(type((1,2,3)))
```

## 1.3 mapping 타입

```python
print(type({'key':'value'}))
```

## 1.4 bool 타입

```python
print(type(True))
```

## 1.5 set 타입

```python
print(type({1,2,3}))
```

# 자료형 확인 실습 01
[문제] 각 데이터의 데이터타입(자료형)을 확인해주세요
1. 15827
2. 7.0
3. {1: 'a'}
4. "Hi, My name is 'Daisy'"
5. (812.324992, 'orange', ('test'))
6. ['apple', 3.513, {'test': 1}]

```python
print('1.', type(15827))
print('2.', type(7.0))
print('3.', type({1: 'a'}))
print('4.', type("Hi, My name is 'Jinseong'"))
print('5.', type((812.324992, 'orange', ('test')))) 
print('6.', type(['apple', 3.513, {'test': 1}]))
```
