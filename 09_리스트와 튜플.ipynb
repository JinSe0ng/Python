# 실습 04
[문제]
사용자로부터 값을 입력 받습니다. \
국어, 영어, 수학, 과학 점수를 한 번에 입력 받습니다. \
이를 kor, eng, mat, sci라는 변수에 각각 저장하고, 사칙 연산을 활용하여 평균 점수를 산출해보세요 \

[실행 결과]
- 국어, 영어, 수학, 과학 점수를 입력해주세요: 98,100,96,100
- 평균 점수는 98.5입니다.


```python
a = input('국어, 영어, 수학, 과학 점수를 입력해주세요: ')

kor,eng,mat,sci = map(int, a.split(','))
b = (kor + eng + mat + sci)/4
print(f'평균 점수는 {b}입니다.' )
```
